```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
rank_data = {
    "name":   ["Liam", "Olivia", "Noah", "Emma", "Oliver", "Ava", "Elijah", "Sophia"],
    "age":    [25, 29, 22, 28, 33, 27, 24, 31],
    "gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "rank":   [1, 2, 3, 4, 5, 6, 7, 8]
}

rank_df = pd.DataFrame(data=rank_data)
rank_df
```

# Matpotlib

```python
plt.bar( rank_df["name"] , rank_df["age"] )

plt.title("Compare Age Vs. Name")
plt.xlabel("Name")
plt.ylabel("Age")

plt.show()
```

```python
plt.plot( rank_df["name"] , rank_df["age"] )

plt.title("Compare Age Vs. Name")
plt.xlabel("Name")
plt.ylabel("Age")

plt.show()
```

# Seaborn

Note :- In Seaborn we use --plt.show()--   for display visuals or 
without plt.show() VS code shows visuals  

```python
plot = sns.barplot(data=rank_df , x = "name" , y = "age")

plot.set_title = "Name Vs. Age"

plt.show()
```

```python
plot = sns.lineplot(data=rank_df , x = "name" , y = "age")

plot.set_title = "Name Vs. Age"

plt.show()
```
