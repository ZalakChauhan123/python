```python
import pandas as pd
```

```python
E_data = {
    "Department": ["HR", "HR", "IT", "IT", "Finance", "Finance","Marketing","Sales","Mfg"],
    "Total Employee": [5, 25, 65, 80, 52, 48,34,23,59],
    "Salary": [4000, 4200, 5000, 5500, 6000, 5800,24454,6543,4244],
    "Bonus": [200, 250, 300, 400, 500, 450,766,7889,4356]
}

EmpDf = pd.DataFrame(E_data)
EmpDf
```

```python
# Give me Average salary of each department
EmpDf.groupby("Department")["Salary"].agg("mean")
```

```python
# Give avg. salary & avg. bonus of each department

EmpDf.groupby("Department").agg({
    "Salary":"mean",
    "Bonus":"mean"
})
# NOTE : We can get what specific we want from specific column...Which we type in agg() function 
```

```python
# How many employees in each department
EmpDf.groupby("Department")["Total Employee"].agg("sum")
```

```python
# In each department find Max. & Min. Salary and Bonus
EmpDf.groupby("Department").agg({
    "Salary":["max","min"],
    "Bonus":["max","min"]
})
```
