```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame
```

```python
dfObj1 = DataFrame(np.arange(36).reshape(6,6))
dfObj1
```

```python
obj2_df = DataFrame(np.arange(15).reshape(5,3))
obj2_df
```

# Concatination of DataFrame

```python
sum_obj1_obj2 = pd.concat(
    [dfObj1, obj2_df]
)
sum_obj1_obj2

# Default It concatinate by Row
```

```python
# Now , in above data frame index is not proper , to fix it

sum_obj1_obj2 = pd.concat(
    
    [dfObj1, obj2_df],
    ignore_index = True
)
sum_obj1_obj2
```

```python
sum_obj1_obj2_column = pd.concat(

    [dfObj1, obj2_df], 
    axis = 1
)
sum_obj1_obj2_column

# Now, It concatinate by Column , default by Row
```
