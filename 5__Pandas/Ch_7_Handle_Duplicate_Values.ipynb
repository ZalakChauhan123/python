```python
import numpy as np
import pandas as pd
from pandas import DataFrame

data = {
    "name": ["Alice", "Bob", "Alice", "Charlie", "Bob", "Charlie", "Alice"],
    "age":  [25, 30, 25, 40, 30, 25, 30],
    "city": ["NY", "LA", "NY", "SF", "LA", "LA", "LA"]
}

obj_df = DataFrame(data)
obj_df
```

```python
# To Check for each Row is there any duplicate value or not
"""
It check all 3 column 
like [Alice	25	NY] is at Row 0 = False & again it repet at Row 2 , so it's True

if duplicate values found then True else False
"""
obj_df.duplicated()
```

```python
# To Drop duplicate values

obj_df.drop_duplicates()
```

```python
# To Drop duplicate values on a specific column
obj_df.drop_duplicates(subset="city")
```
