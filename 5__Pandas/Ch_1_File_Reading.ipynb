```python
# By loc() function - is a label-based data selection method in pandas. It allows you to slice and dice your DataFrame using row labels, column labels

# df.loc[row_indexer, column_indexer]
# row_indexer = Specifies which rows to select — can be a single label, list/slice of labels, boolean array, or even a callable.
# column_indexer = Specifies which columns to select — again, can be labels, lists, slices, boolean arrays, or callables.

subset2 = read_csv.loc[:, ["First Name", "Email"]]
subset2
```
