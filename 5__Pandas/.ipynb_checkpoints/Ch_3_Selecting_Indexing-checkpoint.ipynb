## Indexing

```python
# Set specific Column as Index column of dataset

# Option-1 -- During Read files
excel_data = pd.read_excel(r"C:\Users\zalak\Downloads\student (1).xlsx" , index_col="name")
excel_data.head(3)
```

```python
# For Reset Indexes
excel_data.reset_index(inplace=True)
excel_data.head(3)
```

```python
# Option-2 -- For Set Index (MultiIndexing)
excel_data.set_index(["name","gender"], inplace=True)
excel_data.head()
```

```python
# Create data table
Countries = {
    "Continent":["North America", "Asia", "Europe", "Asia", "Europe","North America","South America","South America", "Asia Minor","Africa","Africa", "Africa"],
    "country":["Peru","India","Germany","China","Sweden","Chili","USA","Canada","Oman","Sudan",'South Africa',"Dijibuti"],
    "Rank":[2,5,8,6,87,24,32,45,12,9,25,11],
    "gdp":[2666,43098,494043,38935,39808,3219807,2783,43958,349913,34808,31489,14908]
}
con_df = pd.DataFrame(data=Countries)
con_df.head()
```

```python
# loc(Row_index, Column_Index) = Label based indexing

# find entry of 2 row
con_df.loc[2]
```

```python
# Find only name & id of all entries
con_df.loc[:,["country","Rank"]]
```

```python
# find name of Top 3 entries
con_df.loc[0:2]
```

```python
# Create multi indexing
con_df.set_index(["Continent","country"], inplace=True)
con_df.head()
```

```python
# In Continent column some of the values repeating itself
con_df.sort_values(["Continent","country"], ascending=[False,True], inplace=True)
con_df
```

```python
# Find out data in multiIndexing
# loc() function used
subset1 = con_df.loc["Asia",:]
subset1
```
