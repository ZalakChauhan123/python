# Change Specific Column name or multiple column names

```python
read_csv = pd.read_csv(r"C:\Users\zalak\Downloads\customers-1000.csv")
read_csv.head(2)
```

```python
# Single Column Rename

modified_df = read_csv.rename( columns = { "Customer Id" : "c_id" } )
modified_df.head(2)
```

```python
# Multiple Column Rename

read_csv.rename( columns = { "First Name":"f_name" , "Last Name":"l_name" , "Country":"Nation" } , inplace = True )
read_csv.columns
```

```python
# Modify all column names
read_csv.columns = ["index","cus_id" , "firstName" , "lastName" , "Organization" , "city" ,"Country" , "Phone" , "Telephone" , "Mail" , "Starting_Date" , "site"]
read_csv.head(2)
```

# Drop Specific or Multiple Columns

```python
# Drop Single Column
dropSingleCol_df = read_csv.drop(columns = ["index"])
dropSingleCol_df.head(2)
```

```python
# Drop Multiple Column
dropMultiCol_df = read_csv.drop( columns = ["site" , "Organization" , "Starting_Date"] )
dropMultiCol_df.head(2)
```
