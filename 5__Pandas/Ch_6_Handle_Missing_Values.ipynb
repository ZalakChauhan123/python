```python
import numpy as np
import pandas as pd
from pandas import DataFrame

rank = {
    "name":   ["Liam", "Olivia", "Noah", "Emma", "Oliver", "Ava", "Elijah", "Sophia"],
    "age":    [25,  29,     22,    28,    33,      27,    24,      31],
    "gender": ["Male","Female","Male","Female","Male","Female","Male","Female"],
    "rank": [1,2,3,4,5,6,7,8]
}

rank_df = DataFrame(rank)
rank_df
```

```python
rank_df.iloc[1:3, 0] = np.nan
rank_df.iloc[4:6, 1] = np.nan
rank_df.iloc[0:4, 2] = np.nan
rank_df.iloc[3,:] = np.nan

rank_df
```

```python
# Check for null values
rank_df.isnull()

# There is another method for find null values
# rank_df.notnul()
```

```python
# fetch out rows based on Null value of specific column

nullAge_df = pd.isnull(rank_df["age"])  # It's called Mask object
rank_df[nullAge_df]
```

# Fill missing values

```python
rank_df.fillna(method="pad")
# Fill previous cell value in NaN cell
```

```python
rank_df.fillna(method="bfill")
# Fill below cell value in NaN cell
```

```python
rank_df.interpolate(method="linear")
```

```python
# dropna() = drop all the cell which has NaN value or It can be said give data frame with No NaN values
rank_df.dropna()
```
