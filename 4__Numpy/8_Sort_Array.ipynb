## Sort() method is used for sorting an array
    This method return copy of the original array
    Original array remains unchanged

```python
import numpy as np
from numpy import random as rand


# Sorting in 1D Array
arr1 = np.array([20,30,44,7,0,42,56])

result1 = np.sort(arr1) # <<------ IMPORTANT LINE

print(result1)




# Sorting in 2D Array
arr2 = rand.randint(100 , size=(3,3))

result2 = np.sort(arr2) # <<------ IMPORTANT LINE

print(result2)



```
