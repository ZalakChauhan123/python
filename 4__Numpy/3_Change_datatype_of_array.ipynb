## Converting Data type of existing Array

-- The best way to change the data type of an existing array, is to make a copy of the array with the astype() method.

-- The astype() function creates a copy of the array, and allows you to specify the data type as a parameter.



```python
import numpy as np

# Convert the data type from int to str
a = [1,2,3,4,5]

arr = np.array(a)
new_arr = arr.astype("str")  # <<---- Important line

print(arr)
print(new_arr)
print(new_arr.dtype)
```

```python
# Convert the data type from str to int

b = ['1', '2', '3', '4', '5']

arr = np.array(b)
new_arr = arr.astype("int") # <<---- Important line

print(arr)
print(new_arr)
print(new_arr.dtype)
```

## Check Data type of existing Array

```python
a = [1.1, 2.1, 3.1, 4.1, 5.1]
arr = np.array(a)
print(arr)
print(arr.dtype)
```
