### Addition , Subtraction & Multiplication 

```python
import numpy as np
from numpy import random as rd

arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([6, 7, 8, 9, 10])

# 1D Array :- Perform Addition method 1
arr3 = np.add(arr1, arr2)
print(arr3)


# 1D Array :- Perform Addition method 2
arr4 = arr1 + arr2
print(arr4)




print()
# Similarly , Perform Subtraction, Multiplication and Division...




# 2D Array :- Perform Addition
arr5 = rd.randint(10 , size=(2,2))
print(arr5)

print()

arr6 = rd.randint(10 , size=(2,2))
print(arr6)

print()

arr7_result = np.add(arr5, arr6)
print("Result 2D Array Addition - \n",arr7_result)

```

## Find out Min & Max values from -- 1D Array --

```python
# Find out Min and Max value from the 1 Dimention array

import numpy as np

arr1d = np.array([1, 2, 3, 4, 5, 20,47,43,56])
print(arr1d) 

# Find out Min value
print("Min Value - ",np.min(arr1d))

# Find out Max value
print("Min Value - ",np.max(arr1d))

# Find Index value of Min and Max value
print("Min Position - ",np.argmin(arr1d))
print("Min Position - ",np.argmax(arr1d))


# Find out sqrt of Array
print("Sqrt of each element - ",np.sqrt(arr1d))
```

## Find out Min & Max values from 2D Array

```python
arr2d = np.array([[11, 254, 365, 478, 5], [666, 780, 878, 9, 1850]])

print(np.min(arr2d)) # Find out Min value from the entire array 
print(np.min(arr2d , axis=0)) # Find out Min value from each row
print(np.min(arr2d , axis=1)) # Find out Min value from each column
```

### shuffle() :- It shuffle the values inside given array
### unique :- It finout unique values inside given array 
            - It is possible to find index number & count of specific value

```python
import numpy as np
from numpy import random as rand

arr1 = rand.randint(100,size=(8))
print("Original Array - ",arr1)


# Shuffle()
shufflled_arr = rand.shuffle(arr1)
print("Shuffled Array - ",arr1)


# Unique()
arr2 = np.array([1,1,3,3,4,4,4,5,6])
print("Original Array - ",arr2)

unique_value_arr = np.unique(arr2, return_index=True, return_counts=True)

print("Unique Value Array - ",unique_value_arr)
```

### Resize() - It changes size of given array

```python
import numpy as np
from numpy import random as rand

arr1 = rand.randint(100, size=(12))
print("Original Array - ",arr1)
print("Dimention = ",arr1.ndim)

print()

# Resize to 2 dimention array 
resized_arr_2d = np.resize(arr1, (2, 6))
print("Resized 2D Array - \n",resized_arr_2d)
print("Dimention = ",resized_arr_2d.ndim)

print()

# Resize to 3 dimention array 
resized_arr_3d = np.resize(arr1, (2,2,3))
print("Resized 3D Array - \n",resized_arr_3d)
print("Dimention = ",resized_arr_3d.ndim)

```

### Insert() & Delete() operation in Array

Insertion :- Possible by insert() & append() method

    -- insert() :- Allows inserting an element at any specific index.
Takes three arguments: array, index, and value.

    -- append() :- Only appends elements at the end of the array.
Takes two arguments: array and values.


```python
import numpy as np
from numpy import random as rand

arr = rand.randint(100, size=(5))
print("Original - ", arr)



# Insert Operation - Single element
updated_arr_insert = np.insert(arr , 0 , 555 )
print("Update_insert - ",updated_arr_insert)

# NOTE :- It is possible to add multiple elements, 
# for example np.insert(arr , 0 , [555, 666, 777] )



# Append Operation 
updated_arr_append = np.append(arr, 1234)
print("Update_append - ",updated_arr_append)

# NOTE :- It is possible to add multiple elements, 
# for example np.append(arr , 0 , [555, 666, 777] )



# Delete Operation
delete_arr = np.delete(arr , 1) # Delete element at index 1
print("Delete at specific index - ",delete_arr)

# Delete last element of the array
delete_lastElement_arr = np.delete( arr , -1 )
print("Delete Last Element - ",delete_lastElement_arr)
```
