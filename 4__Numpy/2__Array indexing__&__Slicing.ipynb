## Array Indexing - Access Specific Element based in index number

```python
# Importing the numpy module
import numpy as np
from numpy import random


arr_1d = random.randint(100, size=(10))
print(arr_1d)


# Accessing elements in a 1D array
print(arr_1d[1])


# Perform Slicing
print(arr_1d[1:3])
print(arr_1d[2:])
print(arr_1d[:5])
```

```python
# Accessing elements in a 2D array

arr_2d = random.randint(200 , size=(3,5))
print(arr_2d)


# Accesing elements in a 2D array
print(arr_2d[0,2])


# Perform Slicing
print(arr_2d[1, 1:3])  # Accessing elements in the second row from the second to the third element
print(arr_2d[0, :])    # Accessing all elements in the first row
print(arr_2d[2, :3])  # Accessing elements in the third row from the first to the third element
```

```python
from numpy import random    

# Accessing Element in a 3D array

arr_3d = random.randint(300,size=(3,3,3))
print(arr_3d)


# Accessing elements in a 3D array
print(arr_3d[1,0,0])


# Perform Slicing
print("Result :- ",arr_3d[1,0:2])

print(arr_3d[1, 0, 0])
```
