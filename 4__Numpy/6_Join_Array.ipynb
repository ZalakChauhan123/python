# Joining Array

- Joining means putting contents of two or more arrays in a single array.



### Join by Concatenation

```python
import numpy as np

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])

arr3 = np.array( [[1, 2], [3, 4]] )
arr4 = np.array( [[5, 6], [7, 8]] )

# Concatinating 1D array
result_1 = np.concatenate((arr1, arr2))

# Concatinating 2D array
result_2 = np.concatenate((arr3, arr4))

print("concatenating 1D array :\n", result_1)
print("concatenating 2D array :\n", result_2)
```

### Join by Stacking

#### stack() function  (1D Array)

```python
import numpy as np

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])

r1 = np.stack((arr1, arr2)) # It becomes 2D Array
r3 = np.stack((arr1, arr2), axis=1) # It becomes 2D Array

print(r1)
print("Dimention of r1 = ", r1.ndim)
print("\n---------------\n", r3)
```

#### stack() function  (2D Array)

```python
import numpy as np

arr3 = np.array( [[1, 2, 3], [4, 5, 6]] )
arr4 = np.array( [[7, 8, 9], [10, 11, 12]] )

r2 = np.stack((arr3, arr4))
r4 = np.stack((arr3, arr4), axis=1)

print(r2)
print("\n---------------\n", r4)
```

### hstack() :- It stack along rows


```python
import numpy as np

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])

arr3 = np.array( [[1, 2, 3], [4, 5, 6]] )
arr4 = np.array( [[7, 8, 9], [10, 11, 12]] )

r1 = np.hstack((arr1, arr2))
r4 = np.hstack((arr3, arr4))


print("\n---------------\n", r1)
print("\n---------------\n", r4)

```

### vstack() :- It stack along columns

```python
import numpy as np

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])
arr3 = np.array( [[1, 2, 3], [4, 5, 6]] )
arr4 = np.array( [[7, 8, 9], [10, 11, 12]] )

r1 = np.vstack((arr1, arr2))
r4 = np.vstack((arr3, arr4))


print("\n---------------\n", r1)
print("\n---------------\n", r4)
```
