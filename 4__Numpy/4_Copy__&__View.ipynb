## Copy in Numpy 
- Copying an array. Changes made in the copy array won't affect the original array

```python
import numpy as np

arr1 = np.array([1,2,3,4,5])

# Create Copy of the array
arr_copy = arr1.copy()

# Change the value of the first element in the original array
arr1[0] = 333

# Print the both array
print(arr1)
print(arr_copy)
```

# View in Numpy
 -- Changes made in the view array will affect the original array

```python
arr2 = np.array([1,2,3,4,5])

# Create a view of the array
arr_view = arr2.view()

# Change the value of the first element in the original array
arr2[0] = 100

# Print the both arrays
print(arr2)
print(arr_view)
```
