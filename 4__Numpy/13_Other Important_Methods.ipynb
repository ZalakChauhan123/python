### arange()
    It Creates array of sequence of numbers with specified step size
    Syntax :- np.arange(start, stop, step, dtype)


```python
import numpy as np

# np.arange(start, stop, step)
arr = np.arange(1, 20, 3)
print(arr)
```

## linspace() :- Creating a Range with Fixed Intervals
    Syntax :- np.linspace(start , stop , num (The number of elements) , endpoint , dtype)


```python
import numpy as np

# np.linspace(start, stop, num (The number of elements), dtype)
arr = np.linspace(1, 20, 5, dtype=int)  # 5 evenly spaced Integer numbers from 1 to 20
print(arr)
```

### full()

    Creates a array with constant values
    Syntax :- np.full(shape, fill_value, dtype)

```python
import numpy as np

# 1D Array
arr_1d = np.full( (10), 5)
print(arr_1d)

arr_1d_float = np.full( (8) , 0.5)
print(arr_1d_float)

arr_1d_str = np.full( (5), "a")
print(arr_1d_str)


print()


# 2D Array
arr_2d = np.full( (3,3), 10)
print(arr_2d)


print()


# 3D Array
arr_3d = np.full ( (2,2,2), 8)
print(arr_3d)


print()


arr = np.full( (5) ,"a")
print(arr)
```
